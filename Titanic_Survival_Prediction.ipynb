```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

```python
# Basic information about the Titanic dataset

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Clean the missing values

# Fill missing Age values with the median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with the most common value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop Cabin because it has too many missing values
df = df.drop('Cabin', axis=1)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
import pandas as pd

# Load the uploaded Titanic dataset
df = pd.read_csv('Titanic-Dataset.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Check the basic information about our dataset

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Handle missing values

# Fill missing Age values with the median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with the most common value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Remove Cabin because it contains many missing values
df = df.drop('Cabin', axis=1)

# Check missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Visualize the number of passengers who survived and did not survive

plt.figure(figsize=(6, 4))

sns.countplot(x='Survived', data=df)

plt.title('Titanic Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
# Survival based on gender

plt.figure(figsize=(6, 4))

sns.countplot(x='Sex', hue='Survived', data=df)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
# Convert gender into numerical values

df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

print("Gender converted successfully!")
print(df[['Sex', 'Survived']].head())
```

```python
# Select features and target

features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']

X = df[features]
y = df['Survived']

print("Input features:")
print(features)

print("\nTarget variable: Survived")
print("\nFirst 5 rows:")
print(X.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create the model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
# Make predictions on the test data

y_pred = model.predict(X_test)

print("Predictions made successfully!")
print(y_pred[:20])
```

```python
# Make predictions on the test data

y_pred = model.predict(X_test)

print("Predictions made successfully!")
print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy: {:.2f}%".format(accuracy * 100))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Create confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

# Display the confusion matrix
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Did Not Survive', 'Survived']
)

disp.plot()
plt.title('Titanic Survival - Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report

# Generate classification report
report = classification_report(
    y_test,
    y_pred,
    target_names=['Did Not Survive', 'Survived']
)

print("Classification Report:")
print(report)
```

```python
print("========== TITANIC SURVIVAL PREDICTION ==========")
print(f"Model Used: Logistic Regression")
print(f"Accuracy: {accuracy * 100:.2f}%")
print("Dataset: Titanic-Dataset.csv")
print("Status: Model trained and evaluated successfully.")
```
