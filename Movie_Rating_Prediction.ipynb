```python
import pandas as pd

# Load the IMDb dataset with the correct encoding
df = pd.read_csv('/content/task2/IMDb Movies India.csv', encoding='latin1')

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
# Understand the movie dataset

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Make a copy of the original dataset
df_clean = df.copy()

# Remove rows where the target (Rating) is missing
df_clean = df_clean.dropna(subset=['Rating'])

# Convert Year into a number
df_clean['Year'] = df_clean['Year'].str.extract(r'(\d{4})').astype(float)

# Convert Duration into minutes
df_clean['Duration'] = df_clean['Duration'].str.extract(r'(\d+)').astype(float)

# Convert Votes into numbers
df_clean['Votes'] = df_clean['Votes'].str.replace(',', '', regex=False)
df_clean['Votes'] = pd.to_numeric(df_clean['Votes'], errors='coerce')

# Fill missing numerical values with their median
df_clean['Year'] = df_clean['Year'].fillna(df_clean['Year'].median())
df_clean['Duration'] = df_clean['Duration'].fillna(df_clean['Duration'].median())
df_clean['Votes'] = df_clean['Votes'].fillna(df_clean['Votes'].median())

# Fill missing categorical values
for column in ['Genre', 'Director', 'Actor 1', 'Actor 2', 'Actor 3']:
    df_clean[column] = df_clean[column].fillna('Unknown')

print("Data cleaning completed!")
print("Rows and columns after cleaning:", df_clean.shape)

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge

# Numerical features
numeric_features = ['Year', 'Duration', 'Votes']

# Categorical features
categorical_features = ['Genre', 'Director', 'Actor 1', 'Actor 2', 'Actor 3']

# Input features and target
X = df_clean[numeric_features + categorical_features]
y = df_clean['Rating']

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

# Create model pipeline
model_improved = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('regressor', Ridge(alpha=10))
    ]
)

print("Feature engineering setup completed!")
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Train the improved model
model_improved.fit(X_train, y_train)

print("Improved model trained successfully!")
```

```python
# Make predictions using the improved model
y_pred_improved = model_improved.predict(X_test)

print("Predictions generated successfully!")
print("First 10 predicted ratings:")
print(y_pred_improved[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_improved = mean_absolute_error(y_test, y_pred_improved)
rmse_improved = np.sqrt(mean_squared_error(y_test, y_pred_improved))
r2_improved = r2_score(y_test, y_pred_improved)

print("========== IMPROVED MODEL EVALUATION ==========")
print(f"Mean Absolute Error (MAE): {mae_improved:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse_improved:.2f}")
print(f"R² Score: {r2_improved:.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred_improved, alpha=0.5)

plt.xlabel('Actual Rating')
plt.ylabel('Predicted Rating')
plt.title('Actual vs Predicted Movie Ratings - Improved Model')

plt.show()
```
